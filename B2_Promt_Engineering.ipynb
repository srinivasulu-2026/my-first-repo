```python
from google.colab import userdata
openai_key = userdata.get('OPENAI_API_KEY')
print(openai_key[:10])
```

```python
from openai import OpenAI
client = OpenAI(api_key = openai_key)
```

```python
response = client.responses.create(
    model="gpt-4o-mini",
    input="What are the foods that I can take for cutting phase on a daily baiss?"
)
print(response.output_text)
```

```python
response = client.responses.create(
    model="gpt-5.6",
    input="I want to become Forward Deployed Engineer in next 6 months, give me the roadmap"
)
print(response.output_text)
```
